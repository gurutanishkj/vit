```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json

# 1. Load Dataset
df = pd.read_csv('../data/creditcard.csv')
print(f"Dataset Shape: {df.shape[0]:,} rows, {df.shape[1]} columns")
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# 2. Class Distribution Analysis
counts = df['Class'].value_counts()
print("Class Distribution:")
print(f"Legitimate (0): {counts[0]:,} ({counts[0]/len(df)*100:.4f}%)")
print(f"Fraud (1):      {counts[1]:,} ({counts[1]/len(df)*100:.4f}%)")

plt.figure(figsize=(6, 4))
counts.plot(kind='bar', color=['#2563eb', '#ef4444'])
plt.title('Severe Class Imbalance in Financial Fraud')
plt.xlabel('Class (0 = Legit, 1 = Fraud)')
plt.ylabel('Transaction Count')
plt.yscale('log')
plt.show()
```

```python
# 3. Amount Distribution Comparison
fraud_amt = df[df['Class'] == 1]['Amount']
legit_amt = df[df['Class'] == 0]['Amount']

print(f"Fraud Amount: Mean=${fraud_amt.mean():.2f}, Median=${fraud_amt.median():.2f}, Max=${fraud_amt.max():.2f}")
print(f"Legit Amount: Mean=${legit_amt.mean():.2f}, Median=${legit_amt.median():.2f}, Max=${legit_amt.max():.2f}")
```
